## Project Title
AI-Based News Article Summarizer using Transformer Models

## Problem Statement
Online news articles are often long and time-consuming to read. 
This project aims to automatically generate concise summaries of 
news articles using transformer-based summarization models.

## Objective
To build a summarization system that converts long news articles 
into short, meaningful summaries using Hugging Face pipelines.

## Model Used
- sshleifer/distilbart-cnn-12-6

## Why This Model
DistilBART is optimized for fast and efficient summarization while 
maintaining good accuracy, making it suitable for CPU-based systems.


```python
from transformers import pipeline
```

```python
summarizer = pipeline(
    "summarization",
    model="sshleifer/distilbart-cnn-12-6"
)
```

```python
article_text = """
Artificial Intelligence is rapidly transforming industries such as healthcare,
finance, transportation, and education. Machine learning models are now capable
of processing massive volumes of data and generating human-like text. Large
language models have improved applications such as chatbots, virtual assistants,
and automated content generation. However, the rapid adoption of AI has raised
serious concerns regarding data privacy, bias, hallucinations, and ethical misuse.
Governments across the world are working on regulatory frameworks to ensure
responsible AI development. Organizations are also investing in AI governance
policies to improve transparency and accountability. Despite these challenges,
AI continues to drive innovation and productivity across multiple sectors.
"""
```

```python
summary = summarizer(
    article_text,
    max_length=80,
    min_length=40,
    do_sample=False
)

print("Summary:")
print(summary[0]["summary_text"])
```

## Observations
- The model successfully condensed the original article into a short summary.
- Key ideas such as AI growth, applications, and ethical concerns were preserved.
- The output is coherent and readable.

## Limitations
- The model may miss very fine details.
- It can sometimes rephrase content instead of strictly extracting facts.
- Performance depends on the quality and length of input text.


## Conclusion
This project demonstrates how transformer-based encoder–decoder models
can be effectively used for text summarization. By leveraging Hugging Face
pipelines, complex NLP tasks can be implemented with minimal code while
achieving meaningful real-world applications.
